```python
import ast
# first genre only (genres is a string like "['drama', 'crime']")
# FIXED: eval() is unsafe, so ast.literal_eval() is used
df['main_genre'] = df['genres'].apply(lambda x: ast.literal_eval(x)[0] if ast.literal_eval(x) else 'unknown')
df['main_country'] = df['production_countries'].apply(lambda x: ast.literal_eval(x)[0] if ast.literal_eval(x) else 'unknown')
```

```python
df_model = pd.get_dummies(df[['type', 'main_genre', 'main_country']], drop_first=True)
df_model = pd.concat([df[['release_year', 'runtime', 'seasons',
                            'imdb_votes', 'tmdb_popularity', 'tmdb_score']], df_model], axis=1)
y = df['imdb_score']
X = df_model
```

```python
y
```

```python
X = X.astype(float)
```

```python
X
```

```python
y = (df['imdb_score'] >= 7).astype(int)
```

```python
import sklearn
from sklearn.model_selection import train_test_split, StratifiedGroupKFold
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
```

```python
# FIXED: the same title appears in many rows (one per actor/director).
# A random split puts the same title in both train and test (data leakage), so we split by title id.
groups = df['id']
sgkf = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)   # one fold = 20% test, class ratio kept
train_idx, test_idx = next(sgkf.split(X, y, groups=groups))
X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

print("Titles present in both train and test:", len(set(groups.iloc[train_idx]) & set(groups.iloc[test_idx])))  # must be 0
```

```python
# FIXED: missing values are filled using TRAIN data only (median)
imputer=SimpleImputer(strategy='median')
X_train=pd.DataFrame(imputer.fit_transform(X_train),columns=X.columns,index=X_train.index)
X_test=pd.DataFrame(imputer.transform(X_test),columns=X.columns,index=X_test.index)

scaler=StandardScaler()
X_train_s=scaler.fit_transform(X_train)
X_test_s=scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, classification_report, confusion_matrix
)
import xgboost as xgb
```

# Logestic Regression Model


```python
# IMPROVED: class_weight='balanced' because good titles (score>=7) are fewer, so recall was low
log_model = LogisticRegression(max_iter=1000, class_weight='balanced')
log_model.fit(X_train_s, y_train)
log_pred = log_model.predict(X_test_s)
```

```python
print("Accuracy :", accuracy_score(y_test, log_pred))
print("Precision:", precision_score(y_test, log_pred))
print("Recall   :", recall_score(y_test, log_pred))
print("F1 Score :", f1_score(y_test, log_pred))

# ROC-AUC needs probabilities, not predicted classes
log_prob = log_model.predict_proba(X_test_s)[:, 1]
print("ROC-AUC  :", roc_auc_score(y_test, log_prob))

print("\nClassification Report:\n", classification_report(y_test, log_pred))

# Confusion matrix plot
cm = confusion_matrix(y_test, log_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix - Logistic Regression')
plt.show()
```

# XGB Boost Model

```python
xgb_model = xgb.XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    scale_pos_weight=(y_train==0).sum()/(y_train==1).sum(),   # IMPROVED: handles class imbalance
    random_state=42,
    eval_metric='logloss'
)
xgb_model.fit(X_train, y_train)   # XGBoost doesn't need scaled data
xgb_pred = xgb_model.predict(X_test)
xgb_prob = xgb_model.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy :", accuracy_score(y_test, xgb_pred))
print("Precision:", precision_score(y_test, xgb_pred))
print("Recall   :", recall_score(y_test, xgb_pred))
print("F1 Score :", f1_score(y_test, xgb_pred))

# ROC-AUC needs probabilities, not predicted classes
# FIXED: use X_test (not X_test_s). The model was trained on unscaled data.
xgb_prob = xgb_model.predict_proba(X_test)[:, 1]
print("ROC-AUC  :", roc_auc_score(y_test, xgb_prob))

print("\nClassification Report:\n", classification_report(y_test, xgb_pred))

# Confusion matrix plot
cm = confusion_matrix(y_test, xgb_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix - XGBoost')
plt.show()
```

# Extra checks (improvements)

```python
# 1. Cross-validation on the training set - more reliable than a single split.
# IMPORTANT: normal KFold/StratifiedKFold is NOT used here, because the same title appears in many rows.
# With normal CV the same title can be in both the train fold and the validation fold (leakage),
# so StratifiedGroupKFold is used. All rows of one title always stay in the same fold.
from sklearn.model_selection import cross_val_score

cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)
groups_train = groups.iloc[train_idx]

# Logistic Regression (uses scaled data)
log_cv = cross_val_score(LogisticRegression(max_iter=1000, class_weight='balanced'),
                         X_train_s, y_train, cv=cv, groups=groups_train, scoring='roc_auc')

# XGBoost (uses unscaled data)
xgb_cv = cross_val_score(xgb.XGBClassifier(n_estimators=200, max_depth=6, learning_rate=0.1,
                                           scale_pos_weight=(y_train==0).sum()/(y_train==1).sum(),
                                           random_state=42, eval_metric='logloss'),
                         X_train, y_train, cv=cv, groups=groups_train, scoring='roc_auc')

print("Logistic Regression CV ROC-AUC:", round(log_cv.mean(), 4), "+/-", round(log_cv.std(), 4))
print("XGBoost             CV ROC-AUC:", round(xgb_cv.mean(), 4), "+/-", round(xgb_cv.std(), 4))
```

```python
# 2. Leakage check: tmdb_score is almost the same as imdb_score, and imdb_votes / tmdb_popularity
# are only known after release. Train again WITHOUT them and compare.
leaky = ['tmdb_score', 'imdb_votes', 'tmdb_popularity']
X_nl = X.drop(columns=leaky)
X_train_nl, X_test_nl = X_nl.iloc[train_idx], X_nl.iloc[test_idx]

# same steps as before: fill missing (train median) and scale
imputer_nl = SimpleImputer(strategy='median')
X_train_nl = imputer_nl.fit_transform(X_train_nl)
X_test_nl = imputer_nl.transform(X_test_nl)

scaler_nl = StandardScaler()
X_train_nl_s = scaler_nl.fit_transform(X_train_nl)
X_test_nl_s = scaler_nl.transform(X_test_nl)

# Logistic Regression
log_nl = LogisticRegression(max_iter=1000, class_weight='balanced')
log_nl.fit(X_train_nl_s, y_train)
log_nl_pred = log_nl.predict(X_test_nl_s)
log_nl_prob = log_nl.predict_proba(X_test_nl_s)[:, 1]

# XGBoost
xgb_nl = xgb.XGBClassifier(n_estimators=200, max_depth=6, learning_rate=0.1,
                           scale_pos_weight=(y_train==0).sum()/(y_train==1).sum(),
                           random_state=42, eval_metric='logloss')
xgb_nl.fit(X_train_nl, y_train)
xgb_nl_pred = xgb_nl.predict(X_test_nl)
xgb_nl_prob = xgb_nl.predict_proba(X_test_nl)[:, 1]

results_nl = pd.DataFrame({
    'Model': ['Logistic Regression', 'XGBoost'],
    'Accuracy': [accuracy_score(y_test, log_nl_pred), accuracy_score(y_test, xgb_nl_pred)],
    'F1': [f1_score(y_test, log_nl_pred), f1_score(y_test, xgb_nl_pred)],
    'ROC-AUC': [roc_auc_score(y_test, log_nl_prob), roc_auc_score(y_test, xgb_nl_prob)]
})
print("Results WITHOUT tmdb_score, imdb_votes, tmdb_popularity:")
print(results_nl.round(4))
```

# Conclusion

The goal of this project was to predict whether a movie or show is “good” (IMDb score of 7 or more) using details like release year, runtime, votes, popularity, type, genre and country.

Data: I joined two datasets (credits and titles) using the id column. After cleaning, the data had 118,127 rows but only 8,156 unique titles, because each title appears once for every actor or director.
Then I cleaned the data:
>Removed columns that were not useful.
Dropped rows where imdb_score was missing (the target is never filled).
Filled seasons with 0 (movies have no seasons).
Filled the other missing numbers with the median, using training data only (after the split).
Removed 169 duplicate rows.

Analysis: I used charts to understand the data. I looked at distributions, outliers, top genres and countries, and the relationships between columns using a correlation heatmap. About 23% of the titles are “good” (score of 7 or more), so the data is imbalanced.

Models: I trained two models and compared them. Class imbalance was handled with class weights. The train/test split is done by title id, so the same title is never in both train and test (checked: 0 titles in both).

Results with all features (test set):

                 Model  Accuracy  Precision  Recall      F1  ROC-AUC
    Logistic Regression    0.8297     0.5842  0.8113  0.6793   0.8960
              XGBoost    0.8741     0.6961  0.7694  0.7309   0.9274

Cross-validation (5 folds, grouped by title) ROC-AUC: Logistic Regression 0.8848 (+/- 0.0137), XGBoost 0.9401 (+/- 0.0065). The small standard deviation shows the models are stable.

Leakage check: tmdb_score, imdb_votes and tmdb_popularity are only known after a title is released, and tmdb_score is the most important feature in the first model. So I trained both models again without these three columns:

                 Model  Accuracy      F1  ROC-AUC
    Logistic Regression    0.7366  0.5326   0.7848
              XGBoost    0.7584  0.5654   0.8078

Result: XGBoost performed better than Logistic Regression in every test. This is because XGBoost can learn complex patterns, while Logistic Regression only learns simple straight-line relationships. But the scores dropped clearly (ROC-AUC from 0.93 to 0.81 for XGBoost) when the three post-release columns were removed. This means the first scores were too optimistic, and the second result (ROC-AUC of about 0.81) is the more honest one, because it only uses information that is known before release.

Limitations:

The same title still appears in many rows (one per actor or director), so rows are not fully independent. The split by title id avoids leakage, but the test results are slightly weighted towards titles with a big cast.
All features are title-level (year, runtime, genre, country, seasons). The cast and director information from the credits file is not used by the model.
Only the first genre and the first country of each title are used.
Missing values were filled with the median, and the imputation for cross-validation was done once on the training set, not separately inside each fold.
IMDb score is based on people's opinions, so some part of it cannot be predicted from these columns.

Final line: Using only information available before release, the model can predict fairly well whether a title will be rated as good (ROC-AUC about 0.81), and XGBoost is the better choice for this dataset. Adding features from the cast and directors (for example the number of cast members or the director’s experience) is the best next step to improve it.
